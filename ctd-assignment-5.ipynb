## Task 1: Handling Missing Data

```python
data = pd.read_csv('/kaggle/input/datasets/chancecayaban/code-the-dream-lesson-5/fake_data.csv')
df = pd.DataFrame(data)
```

```python
df1 = df.dropna()
print("df1")
print(df1)
df1.info()
```

```python
df_filled = df.fillna({'Name':'Unkown', 
                       'Age': df['Age'].mean(), 
                       'Salary': df['Salary'].mean(),
                       'Join Date': '2020-01-01'})
df2 = df_filled.dropna()
df2['Age'] = df2['Age'].astype(int)
print(df2)
```
