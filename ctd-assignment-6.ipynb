## Task 3: Practice Pivot TablesTask 3

```python
data = pd.read_csv('/kaggle/input/datasets/salahuddinahmedshuvo/ecommerce-consumer-behavior-analysis-data/Ecommerce_Consumer_Behavior_Analysis_Data.csv')
ecommerce = pd.DataFrame(data)
print(ecommerce.head(5))

ecommerce['Purchase_Amount'] = ecommerce['Purchase_Amount'].str.replace('$','', regex = False)
ecommerce['Purchase_Amount'] = ecommerce['Purchase_Amount'].astype(float)
vals = ecommerce.groupby('Purchase_Amount').agg({'sum'})

buying_patterns = pd.pivot_table(ecommerce, index = 'Purchase_Category', columns = ['Gender', 'Income_Level'], values = 'Purchase_Amount',aggfunc='sum',fill_value=0)
demographics = pd.pivot_table(ecommerce, index = ['Income_Level', 'Education_Level'], columns = 'Marital_Status', values = 'Customer_ID', aggfunc = 'count', fill_value = 0)
print(demographics)
print(buying_patterns)
```

## Task 4: Practice apply()

```python
data = pd.read_csv('/kaggle/input/datasets/samayashar/ai-powered-job-recommendations/job_recommendation_dataset.csv')
jobs = pd.DataFrame(data)
print(jobs.head(10))

def check_salary(row):
    if row['Experience Level'] != 'Entry Level':
        return 'No'
    if row['Salary'] <= 69999:
        return 'No'
    if 'SQL' not in row['Required Skills']:
        return 'No'
    if 'Python' not in row['Required Skills']:
        return 'No'
    else:
        return 'Yes'
        
jobs['Check These Out'] = jobs.apply(check_salary, axis = 1)

my_jobs = jobs.loc[jobs['Check These Out'] == 'Yes']
print(my_jobs.head(10))
```

## Task 5: Merging and Joining DataFrames

```python
df_1_3_merged = pd.merge(df1, df3, on='Name', how='outer', suffixes=['_left','_right'])
df_1_3_merged['Salary'] = df_1_3_merged['Salary'].fillna(15000)
df_1_3_merged['Favorite Color'] = df_1_3_merged['Favorite Color'].fillna('yellow')
df_1_3_merged['Age'] = np.where(df_1_3_merged['Age_left'].notna(),df_1_3_merged['Age_left'], df_1_3_merged['Age_right'])
df_1_3_merged = df_1_3_merged.drop(columns=['Age_left', 'Age_right'])
print(df_1_3_merged)

df1_b = df1.set_index('Name')
df3_b = df3.set_index('Name')
joined = df1_b.join(df3_b, how='outer', lsuffix = '_left', rsuffix = '_right')
joined['Age'] = np.where(joined['Age_left'].notna(),joined['Age_left'], joined['Age_right'])
joined = joined.drop(columns=['Age_left', 'Age_right'])
print(joined)
```

## Task 11: Data Wrangling a Kaggle Dataset

```python
data = pd.read_csv('/kaggle/input/private-dataset/results.csv')
football_results = pd.DataFrame(data)
print(football_results.head())

results1 = football_results[['home_team','away_team','home_score','away_score', 'date']]
print(results1.head())
results2 = results1.rename(columns={'home_team' : 'team',
                                    'away_team' : 'opponent',
                                    'home_score' : 'points_for',
                                    'away_score' : 'points_against'})
print(results2)
results3 = results1.rename(columns={'home_team' : 'opponent',
                                    'away_team' : 'team',
                                    'home_score' : 'points_against',
                                    'away_score' : 'points_for'})
print(results3)

football_results = pd.concat([results2, results3])
football_results = football_results.reset_index(drop=True)
print(football_results.head())

points_against = football_results.groupby('team').agg({'points_against' : 'mean'})
points_against = points_against.sort_values(by = 'points_against', ascending = False)
print(points_against.head(10))
```
