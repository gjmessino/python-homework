```python
data = pd.read_csv('/kaggle/input/datasets/salahuddinahmedshuvo/ecommerce-consumer-behavior-analysis-data/Ecommerce_Consumer_Behavior_Analysis_Data.csv')
ecommerce = pd.DataFrame(data)
# print(ecommerce.head())

ecommerce['Purchase_Amount'] = ecommerce['Purchase_Amount'].str.replace('$','', regex = False)
ecommerce['Purchase_Amount'] = ecommerce['Purchase_Amount'].astype(float)
vals = ecommerce.groupby('Purchase_Amount').agg({'sum'})

buying_patterns = pd.pivot_table(ecommerce, index = 'Purchase_Category', columns = ['Gender', 'Income_Level'], values = 'Purchase_Amount',aggfunc='sum',fill_value=0)
demographics = pd.pivot_table(ecommerce, index = ['Income_Level', 'Education_Level'], columns = 'Marital_Status', values = 'Customer_ID', aggfunc = 'count', fill_value = 0)
print(demographics)
```
